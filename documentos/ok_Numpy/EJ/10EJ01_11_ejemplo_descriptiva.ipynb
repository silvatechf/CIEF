```python
import pandas as pd
ages = [26,26,27,22,22,19,22,26,29,29,27,23,28,19,52]

lista = pd.DataFrame(ages)
media = lista.mean()
moda = lista.mode()
mediana = lista.median()
q1 = lista.quantile(0.25)
q2 = lista.quantile(0.50)
q3 = lista.quantile(0.75)
rango = lista.max()-lista.min()
varianza = lista.var()
desviacion = lista.std()
iqr = q3-q1
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr
outliers = (lista < lower_bound) | (lista > upper_bound)
outliers_values = lista[outliers]
print("Límites inferiores y superiores para detectar outliers:")
print(f"\nLímites inferiores:\n{lower_bound}")
print(f"\nLímites superiores:\n{upper_bound}")

print("\nValores atípicos detectados:")
print(outliers_values)
#Centralización
print('Media: ',media)
print('Moda: ',moda)
print('Mediana: ',mediana)
print('Q1:',q1)
print('Q2:',q2)
print('Q3:',q3)
#Dispersión
print('Rango:',rango)
print('Varianza',varianza)
print('Desviación',desviacion,'-',varianza**0.5)
print('IQR.',iqr)
print(lista.describe())
```
