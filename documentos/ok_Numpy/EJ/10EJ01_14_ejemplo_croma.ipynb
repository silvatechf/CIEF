```python
imagen = cv2.imread('R.jpg')
```

```python
imagen_np = np.array(imagen)
print(imagen_np.shape)
print(imagen)
```

```python
# Convertir la imagen al espacio de color HSV (Hue, Saturation, Value)
hsv = cv2.cvtColor(imagen, cv2.COLOR_BGR2HSV)
```

```python
bajo_verde = np.array([35, 40, 40])
alto_verde = np.array([85, 255, 255])
```

```python
# Crear una máscara con los píxeles que están dentro del rango del color verde
mascara = cv2.inRange(hsv, bajo_verde, alto_verde)
```

```python
# Invertir la máscara para obtener la imagen sin el fondo verde
mascara_invertida = cv2.bitwise_not(mascara)
```

```python
# Aplicar la máscara invertida para quitar el fondo verde de la imagen original
resultado = cv2.bitwise_and(imagen, imagen, mask=mascara_invertida)
```

```python
# Mostrar la imagen original y el resultado sin el fondo verde
cv2.imshow('Imagen Original', imagen)
cv2.imshow('Imagen sin Fondo Verde', resultado)
```

```python
# Guardar la imagen sin el fondo verde
cv2.imwrite('imagen_sin_fondo.jpg', resultado)
```
