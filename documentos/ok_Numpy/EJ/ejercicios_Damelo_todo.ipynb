```python
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt
from scipy.ndimage import sobel

# Cargar la imagen y convertirla a escala de grises
imagen = Image.open('oficina.jpg').convert('L')

# Convertir la imagen a un array de numpy
imagen_array = np.array(imagen)

# Aplicar el filtro de Sobel en la dirección horizontal y vertical
sobel_x = sobel(imagen_array, axis=0)  # Sobel en la dirección X
sobel_y = sobel(imagen_array, axis=1)  # Sobel en la dirección Y

# Magnitud del gradiente para combinar ambos
sobel_total = np.hypot(sobel_x, sobel_y)  # Calcula la magnitud total del gradiente

# Normalizar el resultado
sobel_total = (sobel_total / np.max(sobel_total)) * 255  # Escalar entre 0 y 255

# Mostrar la imagen original y la imagen con el filtro Sobel aplicado
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 5))

ax1.imshow(imagen_array, cmap='gray')
ax1.set_title("Imagen Original")
ax1.axis('off')

ax2.imshow(sobel_total, cmap='gray')
ax2.set_title("Imagen con Filtro Sobel (Detección de Bordes)")
ax2.axis('off')

plt.show()
```

```python
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt

# Función para aplicar el filtro sepia ajustable
def aplicar_filtro_sepia(imagen_array, intensidad=1.0):
    # Crear una copia de la imagen para aplicar el filtro
    sepia = np.copy(imagen_array)
    
    # Definir los coeficientes del filtro sepia
    tr = [0.393, 0.769, 0.189]
    tg = [0.349, 0.686, 0.168]
    tb = [0.272, 0.534, 0.131]
    
    # Aplicar el filtro a cada píxel (R, G, B)
    for i in range(imagen_array.shape[0]):
        for j in range(imagen_array.shape[1]):
            r, g, b = imagen_array[i, j]
            
            # Nuevos valores de rojo, verde y azul
            nuevo_r = tr[0]*r + tr[1]*g + tr[2]*b
            nuevo_g = tg[0]*r + tg[1]*g + tg[2]*b
            nuevo_b = tb[0]*r + tb[1]*g + tb[2]*b
            
            # Aplicar la intensidad del efecto sepia (mezclando con el color original)
            sepia[i, j, 0] = min(255, (intensidad * nuevo_r + (1 - intensidad) * r))
            sepia[i, j, 1] = min(255, (intensidad * nuevo_g + (1 - intensidad) * g))
            sepia[i, j, 2] = min(255, (intensidad * nuevo_b + (1 - intensidad) * b))
    
    return sepia

# Cargar la imagen y convertirla a RGB
imagen = Image.open('oficina.jpg').convert('RGB')

# Convertir la imagen a un array de numpy
imagen_array = np.array(imagen)

# Ajustar la intensidad del filtro sepia (valores entre 0 y 1)
intensidad = 0.8  # 0 es sin sepia, 1 es sepia completo
imagen_sepia = aplicar_filtro_sepia(imagen_array, intensidad=intensidad)

# Convertir de nuevo el array a imagen para mostrarla
imagen_sepia_pil = Image.fromarray(imagen_sepia.astype('uint8'))

# Mostrar la imagen original y la imagen con filtro sepia ajustado
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 5))

ax1.imshow(imagen_array)
ax1.set_title("Imagen Original")
ax1.axis('off')

ax2.imshow(imagen_sepia_pil)
ax2.set_title(f"Imagen con Filtro Sepia (Intensidad {intensidad})")
ax2.axis('off')

plt.show()
```

```python
import cv2
import numpy as np

# Cargar la imagen
imagen = cv2.imread('oficina.jpg')

# Separar los canales de color BGR (Blue, Green, Red)
B, G, R = cv2.split(imagen)

# Establecer el canal azul a 0 (o un valor bajo si quieres reducir el azul)
B.fill(0)


# Recomponer la imagen sin el canal azul
imagen_sin_azul = cv2.merge([B, G, R])

# Guardar la imagen procesada
cv2.imwrite('imagen_sin_azul.jpg', imagen_sin_azul)

# Mostrar la imagen original y la imagen sin azul
cv2.imshow('Imagen Original', imagen)
cv2.imshow('Imagen sin Azul', imagen_sin_azul)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
