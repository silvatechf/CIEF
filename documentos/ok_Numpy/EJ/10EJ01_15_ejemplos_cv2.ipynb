```python
#1. Crear una imagen desde cero con NumPy y dibujar con OpenCV
import cv2
import numpy as np

# Crear una imagen negra (512x512 píxeles, 3 canales RGB)
img = np.zeros((512, 512, 3), dtype=np.uint8)

# Dibujar una línea azul
cv2.line(img, (0, 0), (511, 511), (255, 0, 0), 5)

# Dibujar un rectángulo verde
cv2.rectangle(img, (100, 100), (400, 400), (0, 255, 0), 3)

# Dibujar un círculo rojo
cv2.circle(img, (256, 256), 100, (0, 0, 255), -1)

# Mostrar imagen
cv2.imshow("Imagen creada", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
#2. Cargar una imagen y aplicar efectos con NumPy
import cv2
import numpy as np

# Cargar imagen desde archivo
img = cv2.imread('R.jpg')

# Convertir a escala de grises
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Invertir la imagen (efecto negativo)
inverted = 255 - gray

# Mostrar resultados
cv2.imshow('Original', img)
cv2.imshow('Gris', gray)
cv2.imshow('Negativo', inverted)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# 3. Usar NumPy para modificar una región de la imagen
import cv2
import numpy as np

# Cargar imagen
img = cv2.imread('oficina.jpg')

# Seleccionar una región de interés (ROI)
roi = img[100:200, 200:300]

# Pegar la ROI en otra parte de la imagen
img[0:100, 0:100] = roi

cv2.imshow("Modificada", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
#4. Aplicar filtros con NumPy
import cv2
import numpy as np

# Cargar imagen
img = cv2.imread('R.jpg')

# Convertir a gris
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Umbral manual (todo píxel mayor a 128 se vuelve blanco)
_, thresh = cv2.threshold(gray, 128, 255, cv2.THRESH_BINARY)

cv2.imshow('Original', img)
cv2.imshow('Umbral', thresh)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
